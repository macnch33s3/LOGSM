### Correlation Check

Code Block Nr.: 14

```python
# Choose your aggregation level here
FREQ = "D"    # "D" (daily), "W" (weekly), "ME" (monthly)

# Ensure datetime index
dfc = df_cleaned.copy()
dfc['Datum'] = pd.to_datetime(dfc['Datum'], errors='coerce', dayfirst=True)
dfc = dfc.set_index('Datum').sort_index()

# Aggregate Menge by sum and Temperatur by chosen rule, at selected frequency
df_agg = dfc.resample(FREQ).agg({
    "Menge [hl]": "sum",
    "Temp [C]": "mean"
}).dropna()

# --- Scatter with regression line ---
plt.figure(figsize=(8, 6))
sns.regplot(data=df_agg, x="Temp [C]", y="Menge [hl]", line_kws={"color": "red"})
plt.xlabel('Temperatur (°C)')
plt.ylabel('Menge [hl]')
plt.title(f'Zusammenhang (Aggregation: {FREQ})')
plt.grid(True)
plt.show()

# --- Time-series plot (dual y-axes) ---
plt.figure(figsize=(12, 6))
ax1 = df_agg["Menge [hl]"].plot(label="Menge [hl]", linewidth=2, color="tab:blue")
ax2 = ax1.twinx()
df_agg["Temp [C]"].plot(ax=ax2, label="Temperatur (°C)", linewidth=2, color="tab:red")

ax1.set_ylabel("Menge [hl]", color="tab:blue")
ax2.set_ylabel("Temperatur (°C)", color="tab:orange")
plt.title(f'Zeitreihen (Aggregation: {FREQ})')
ax1.grid(True)

# two legends (one per axis)
lines1, labels1 = ax1.get_legend_handles_labels()
lines2, labels2 = ax2.get_legend_handles_labels()
ax1.legend(lines1 + lines2, labels1 + labels2, loc="upper left")
plt.show()

# --- Correlation + significance ---
corr = df_agg["Menge [hl]"].corr(df_agg["Temp [C]"])
r, p_value = pearsonr(df_agg["Menge [hl]"], df_agg["Temp [C]"])
print(f"Aggregation: {FREQ}")
print(f"  Pearson correlation: {corr:.2f}")
print(f"  p-value: {p_value:.8f}")
```

# Exercise 3 - Forecastability check

Code Block Nr.: 15

```python
# Create working DataFrame
df_beerdemand = df[['Datum', 'Menge [hl]']].copy()

# Ensure datetime index
df_beerdemand['Datum'] = pd.to_datetime(df_beerdemand['Datum'])
df_beerdemand.set_index('Datum', inplace=True)

df_beerdemand
```

Forecastability Check:

1) Plot the time series
2) Calculate the ADI and the CoV to make the classification
3) Plot classification chart

--> Adjust the aggregation level form daily to weekly to monthly

Code Block Nr.: 16

```python
# Weekly Aggregation [D = Daily, W = Weekly, ME = Monthly]
df_beerdemand = df_beerdemand.resample('W').sum()

# Plot time series
plt.figure(figsize=(12, 4))
plt.plot(df_beerdemand.index, df_beerdemand['Menge [hl]'], color='blue', label='Demand')

# Formatting
plt.title('Time Series: Beer Demand')
plt.xlabel('Datum')
plt.ylabel('Demand [hl]')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

Code Block Nr.: 17

```python
# Calculate ADI and CoV
non_zero = df_beerdemand['Menge [hl]'] > 0
adi = len(df_beerdemand) / non_zero.sum()
cov = df_beerdemand['Menge [hl]'].std() / df_beerdemand['Menge [hl]'].mean()

# Print classification values
print(f"ADI (Average Demand Interval): {adi:.2f}")
print(f"CoV (Coefficient of Variation): {cov:.2f}")

# Classify demand pattern
if adi <= 1.32 and cov <= 0.49:
    pattern = "Smooth"
elif adi > 1.32 and cov <= 0.49:
    pattern = "Intermittent"
elif adi <= 1.32 and cov > 0.49:
    pattern = "Erratic"
else:
    pattern = "Lumpy"

print(f"📊 Demand Pattern Classification: {pattern}")
```

Code Block Nr.: 18

```python
# Plot
fig, ax = plt.subplots(figsize=(8, 6))
ax.scatter(cov, adi, color='red', s=300)
ax.text(cov + 0.02, adi + 0.02, 'Menge [hl]', fontsize=12)

# Threshold lines (note the swap)
ax.axvline(0.49, color='gray', linestyle='--')  # CoV threshold
ax.axhline(1.32, color='gray', linestyle='--')  # ADI threshold

# Annotate regions
ax.text(0.05, 0.7, 'Smooth', fontsize=20, color='green')
ax.text(0.8, 0.7, 'Erratic', fontsize=20, color='orange')
ax.text(0.05, 1.6, 'Intermittent', fontsize=20, color='blue')
ax.text(0.8, 1.6, 'Lumpy', fontsize=20, color='purple')

# Ensure axes include the point
ax.set_xlim(0, max(1.5, cov + 0.5))
ax.set_ylim(0, max(2, adi + 0.5))

# Labels and title
ax.set_xlabel('CoV (Coefficient of Variation)')
ax.set_ylabel('ADI (Average Demand Interval)')
ax.set_title('Demand Classification: CoV vs ADI')
plt.grid(True)
plt.show()
```

# Exercise 4 - Statistical Forecasting

### 1. Forecast Preparation

Code Block Nr.: 19

```python
# Exclude first and last row
df_beerdemand_prepared = df_beerdemand.iloc[1:-1].copy()

# Plot time series
plt.figure(figsize=(12, 4))
plt.plot(df_beerdemand_prepared.index, df_beerdemand_prepared['Menge [hl]'], color='blue', label='Demand')

# Formatting
plt.title('Time Series: Beer Demand')
plt.xlabel('Datum')
plt.ylabel('Demand [hl]')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# Count number of datapoints
len(df_beerdemand_prepared)
```

Code Block Nr.: 20

```python
# Define test size
test_size = 24

# Split
df_train = df_beerdemand_prepared.iloc[:-test_size].copy()
df_test = df_beerdemand_prepared.iloc[-test_size:].copy()

df_train
```

### 2. Model Selection

Code Block Nr.: 21

```python
# ETS Forecast
ets_model = ExponentialSmoothing(df_train['Menge [hl]'], trend='add', seasonal='add').fit()
# ets_model = ExponentialSmoothing(df_train['Menge [hl]'], trend='mul', seasonal='mul', seasonal_periods=7, initialization_method='estimated').fit()
ets_forecast = ets_model.forecast(steps=24)

# Simple ARIMA: (p,d,q) can be tuned; try (1,1,1) as a basic start
arima_model = ARIMA(df_train['Menge [hl]'], order=(12,1,12)).fit() 
arima_forecast = arima_model.forecast(steps=24)

# SARIMA: includes seasonality (p,d,q)(P,D,Q,s)
# Example: weekly seasonality s = 7
sarima_model = SARIMAX(df_train['Menge [hl]'], order=(6,1,6), seasonal_order=(2,0,2,7)).fit()
sarima_forecast = sarima_model.forecast(steps=24)

# Moving average
# We'll use the last k values of the training set as the average. Let’s use a 3-day moving average.
k = 2 #24
ma_forecast = pd.Series([df_train['Menge [hl]'].iloc[-k:].mean()] * 24, index=df_test.index)

# Combine all forecasts into one dataframe
# Align all forecasts with df_test index
forecast_df = pd.DataFrame({
    'ETS': ets_forecast.values,
    'ARIMA': arima_forecast.values,
    'SARIMA': sarima_forecast.values,
    'MovingAverage': ma_forecast.values
}, index=df_test.index)


# --- Plot ---
plt.figure(figsize=(14, 6))

# Training data
plt.plot(df_train.index, df_train['Menge [hl]'], label='Train Data', color='gray')

# Test data (actual values)
plt.plot(df_test.index, df_test['Menge [hl]'], label='Test Data (Actual)', color='black', linewidth=2)

# Forecasts
plt.plot(forecast_df.index, forecast_df['ETS'], label='ETS Forecast')
plt.plot(forecast_df.index, forecast_df['ARIMA'], label='ARIMA Forecast')
plt.plot(forecast_df.index, forecast_df['SARIMA'], label='SARIMA Forecast')
plt.plot(forecast_df.index, forecast_df['MovingAverage'], label='Moving Avg Forecast')

# Formatting
plt.title('Forecast vs. Actual (Test Set)')
plt.xlabel('Datum')
plt.ylabel('Menge [hl]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# --- Plot ---
plt.figure(figsize=(14, 6))

# Test data (actual values)
plt.plot(df_test.index, df_test['Menge [hl]'], label='Test Data (Actual)', color='black', linewidth=2)

# Forecasts
plt.plot(forecast_df.index, forecast_df['ETS'], label='ETS Forecast')
plt.plot(forecast_df.index, forecast_df['ARIMA'], label='ARIMA Forecast')
plt.plot(forecast_df.index, forecast_df['SARIMA'], label='SARIMA Forecast')
plt.plot(forecast_df.index, forecast_df['MovingAverage'], label='Moving Avg Forecast')

# Formatting
plt.title('Forecast vs. Actual (Test Set)')
plt.xlabel('Datum')
plt.ylabel('Menge [hl]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# --- 2. MAPE Calculation ---
mape_results = {
    'ETS': mape(df_test['Menge [hl]'], forecast_df['ETS']),
    'ARIMA': mape(df_test['Menge [hl]'], forecast_df['ARIMA']),
    'SARIMA': mape(df_test['Menge [hl]'], forecast_df['SARIMA']),
    'MovingAverage': mape(df_test['Menge [hl]'], forecast_df['MovingAverage']),
}

# Print MAPE results
for model, error in mape_results.items():
    print(f"MAPE for {model}: {error:.2%}")

# --- 2. MAE Calculation ---
mae_results = {
    'ETS': mae(df_test['Menge [hl]'], forecast_df['ETS']),
    'ARIMA': mae(df_test['Menge [hl]'], forecast_df['ARIMA']),
    'SARIMA': mae(df_test['Menge [hl]'], forecast_df['SARIMA']),
    'MovingAverage': mae(df_test['Menge [hl]'], forecast_df['MovingAverage']),
}

# Print MAE results
for model, error in mae_results.items():
    print(f"MAE for {model}: {error:.5}")

# --- 2. MSE Calculation ---
mse_results = {
    'ETS': mse(df_test['Menge [hl]'], forecast_df['ETS']),
    'ARIMA': mse(df_test['Menge [hl]'], forecast_df['ARIMA']),
    'SARIMA': mse(df_test['Menge [hl]'], forecast_df['SARIMA']),
    'MovingAverage': mse(df_test['Menge [hl]'], forecast_df['MovingAverage']),
}

# Print MSE results
for model, error in mse_results.items():
    print(f"MSE for {model}: {error:.5}")
```

# Exercise 5 - ML Forecasting (Prophet)

### 1. Performance Compairson

Code Block Nr.: 23

```python
# Resample df_train to weekly if it's not already
df_prophet_train_weekly = df_train.resample('W').sum()

# Prepare training data
df_prophet_train_weekly = df_prophet_train_weekly.reset_index().rename(columns={'Datum': 'ds', 'Menge [hl]': 'y'})

# Initialize and fit the model
prophet_model = Prophet(daily_seasonality=False, weekly_seasonality=True, yearly_seasonality=True)
prophet_model.fit(df_prophet_train_weekly)

# Create future dataframe for 24 days
future = prophet_model.make_future_dataframe(periods=24, freq='W')

# Forecast
forecast_prophet = prophet_model.predict(future)

# Extract only the forecast part (last 24 rows)
prophet_forecast = forecast_prophet.set_index('ds').loc[df_test.index, 'yhat']

# Add Prothet forecast to forecast dataframe
forecast_df['Prophet'] = prophet_forecast.values

# Add Prophet forecast to chart
plt.figure(figsize=(14, 6))

plt.plot(df_train.index, df_train['Menge [hl]'], label='Train Data', color='gray')
plt.plot(df_test.index, df_test['Menge [hl]'], label='Test Data (Actual)', color='black', linewidth=2)
plt.plot(forecast_df.index, forecast_df['ETS'], label='ETS')
plt.plot(forecast_df.index, forecast_df['ARIMA'], label='ARIMA')
plt.plot(forecast_df.index, forecast_df['SARIMA'], label='SARIMA')
plt.plot(forecast_df.index, forecast_df['MovingAverage'], label='Moving Average')
plt.plot(forecast_df.index, forecast_df['Prophet'], label='Prophet')

plt.title('Forecasts')
plt.xlabel('Datum')
plt.ylabel('Menge [hl]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Add Prophet forecast to chart
plt.figure(figsize=(14, 6))

plt.plot(df_test.index, df_test['Menge [hl]'], label='Test Data (Actual)', color='black', linewidth=2)
plt.plot(forecast_df.index, forecast_df['ETS'], label='ETS')
plt.plot(forecast_df.index, forecast_df['ARIMA'], label='ARIMA')
plt.plot(forecast_df.index, forecast_df['SARIMA'], label='SARIMA')
plt.plot(forecast_df.index, forecast_df['MovingAverage'], label='Moving Average')
plt.plot(forecast_df.index, forecast_df['Prophet'], label='Prophet')

plt.title('Forecasts')
plt.xlabel('Datum')
plt.ylabel('Menge [hl]')
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.show()

# Update MAPE dictionary
mape_results['Prophet'] = mape(df_test['Menge [hl]'], forecast_df['Prophet'])

# Print all MAPE results
for model, error in mape_results.items():
    print(f"MAPE for {model}: {error:.2%}")
```
